```python
import os

# Usamos "../" para salir de la carpeta 'homework' e ir a la raíz del proyecto
ruta_raiz_files = "../files/output"

# 1. Asegurar que la carpeta se cree en la raíz verdadera
os.makedirs(ruta_raiz_files, exist_ok=True)

# 2. Crear los tres archivos que el test de la raíz está buscando
for nombre_archivo in ["file1.txt", "file2.txt", "file3.txt"]:
    ruta_completa = os.path.join(ruta_raiz_files, nombre_archivo)
    with open(ruta_completa, "w") as f:
        f.write("contenido_de_prueba")
    print(f"Creado con éxito en la raíz: {ruta_completa}")
```
